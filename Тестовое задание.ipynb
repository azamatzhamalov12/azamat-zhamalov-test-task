```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_excel(
    "Данные для тестового задания.xlsx",
    sheet_name="Данные АБ тестов"
)

df.head()
```

```python
df['experiment_num'].value_counts().sort_index()
```

```python
arpu = (
    df.groupby(['experiment_num', 'experiment_group'])
      .agg(
          users=('user_id', 'nunique'),
          revenue=('revenue', 'sum')
      )
)

arpu['ARPU'] = arpu['revenue'] / arpu['users']

arpu
```

```python
for exp in sorted(df['experiment_num'].unique()):

    exp_df = df[df['experiment_num'] == exp]

    control = exp_df[
        exp_df['experiment_group'] == 'control'
    ]['revenue']

    test = exp_df[
        exp_df['experiment_group'] == 'test'
    ]['revenue']

    stat, pvalue = stats.ttest_ind(
        control,
        test,
        equal_var=False
    )

    print(f'Experiment {exp}')
    print(f'p-value = {pvalue:.6f}')
    print('-' * 30)
```
